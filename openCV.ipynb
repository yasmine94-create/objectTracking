```python
#  BGR >  RGB 
img_rgb  =  cv2.cvtColor(img ,  cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

```python
# blur
img_blur  =  cv2.GaussianBlur(img_rgb , (3,3) , 0) 
plt.imshow(img_blur)
```

```python
# edge detection 

img_canny = cv2.Canny(img_rgb , 100 , 200 )
plt.imshow(img_canny,  cmap= 'gray')
```

```python
# edge detection 

img_canny = cv2.Canny(img_blur , 150 , 250 )
plt.imshow(img_canny,  cmap= 'gray')
```

```python
#  creat image  >  
blank = np.zeros((400, 400, 3) ,dtype='uint8' )
plt.imshow(blank)
```

```python
cv2.rectangle(blank ,(50, 50) , (150, 150) , (255,0,0), thickness=2) # red
plt.imshow(blank)
```

```python
cv2.rectangle(blank ,(0, 0) , (150, 150) , (255,0,0), thickness=2) # blue
cv2.imshow('blank' , blank)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.circle(blank , (200,200) , 100 , (0,255,0) ,  thickness=2)
plt.imshow(blank)
```

```python
cv2.putText(blank , 'text' , (300, 300),cv2.FONT_HERSHEY_SCRIPT_SIMPLEX ,1, (0,0,255), 2)
plt.imshow(blank)
```

```python
cap =cv2.VideoCapture(r'Data\vtest.avi')
while cap.isOpened() : 
    flag  , frame  = cap.read()

    if not flag :
        break

    cv2.imshow('video' , frame)  

    if  cv2.waitKey(20)  &  0xFF ==ord('d') :
        break 

    if  cv2.getWindowProperty('video' ,cv2.WND_PROP_VISIBLE)  < 1 : 
        break



cap.release()
cv2.destroyAllWindows()    
```
